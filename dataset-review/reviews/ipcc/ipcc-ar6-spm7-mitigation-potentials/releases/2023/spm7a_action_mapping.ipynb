# SPM.7(a) options → city action list — mapping

Links each AR6 SPM.7(a) mitigation option to actions in the city action list
(102 actions, c40/icare/ipcc publishers, from the SR1.5 feasibility review's
`actions_to_sr15_mapping.csv`). Same design as the TEF mapping: **link, not
attribute** — potential stays at option level; actions inherit a pointer and
band, never a split value.

- Inputs: `data/spm7a_mitigation_options_2030_clean.csv` (this release);
  action list from `../../../ipcc-sr15-mitigation-feasibility/releases/2018/data/`
  (committed data, cross-review dependency).
- Output: `data/spm7a_option_to_action.csv`
- Unmapped options kept with `mapping_confidence: none` and rationale.

```python
import pandas as pd

OPTIONS = "data/spm7a_mitigation_options_2030_clean.csv"
ACTIONS = ("../../../ipcc-sr15-mitigation-feasibility/releases/2018/data/"
           "actions_to_sr15_mapping.csv")
OUT = "data/spm7a_option_to_action.csv"

opts = pd.read_csv(OPTIONS)
acts = pd.read_csv(ACTIONS)[["action_id", "action_name", "primary_channel"]]
assert acts.action_id.is_unique
print(f"{len(opts)} options, {len(acts)} actions")
```

## Expand, validate

```python
rows = []
for m in MAPPING:
    if not m["ids"]:
        rows.append(dict(option=m["option"], action_id=None,
                         confidence=m["confidence"], rationale=m["rationale"]))
    for a in m["ids"]:
        rows.append(dict(option=m["option"], action_id=a,
                         confidence=m["confidence"], rationale=m["rationale"]))
draft = pd.DataFrame(rows)

assert set(opts.option) == set(draft.option), set(opts.option) ^ set(draft.option)
mapped = draft.dropna(subset=["action_id"])
unknown = set(mapped.action_id) - set(acts.action_id)
assert not unknown, f"unresolvable action ids: {unknown}"
dup = mapped.duplicated(subset=["option", "action_id"])
assert not dup.any(), mapped[dup]

cov_opt = draft.groupby("option").action_id.apply(lambda s: s.notna().any())
unreached = sorted(set(acts.action_id) - set(mapped.action_id))
print(f"options mapped to >=1 action: {cov_opt.sum()}/{len(cov_opt)}")
print(f"unmapped options: {sorted(cov_opt[~cov_opt].index)}")
print(f"actions reached: {mapped.action_id.nunique()}/{len(acts)}; unreached: {len(unreached)}")
print("unreached action ids:", unreached)
```

## Export

```python
slug = opts.option.str.lower().str.replace(r"[^a-z0-9]+", "_", regex=True).str.strip("_")
opt_id = dict(zip(opts.option, "spm7a_" + slug))

out = mapped.merge(acts, on="action_id")
none_rows = draft[draft.action_id.isna()].assign(action_name="", primary_channel="")
out = pd.concat([out, none_rows], ignore_index=True)
out["src_option_id"] = out.option.map(opt_id)
out["src_option_name"] = out.option
out["action_publisher"] = out.action_id.fillna("").str.split("_").str[0]
out = out.rename(columns={"confidence": "mapping_confidence"})
out = out[["src_option_id", "src_option_name", "action_id", "action_name",
           "action_publisher", "primary_channel", "mapping_confidence", "rationale"]]
out = out.sort_values(["src_option_id", "action_id"], na_position="last").reset_index(drop=True)
out.to_csv(OUT, index=False)
print(f"wrote {OUT}: {len(out)} rows")
out.mapping_confidence.value_counts()
```

## Findings

- **25/31 options map to >=1 action; 89/102 actions reached** (99 rows: 59 high, 34 medium, 6 none).
- **Coverage is a property of the target taxonomy, not the dataset**: vs the TEF mapping (19/31 options, high band -> 4 TEs), this action list is land/agriculture-rich, so the global high band reaches **16 actions** (vs 4). Band inheritance: high -> 16 actions, medium -> 74, low (CCU) -> 1; 2 actions span bands.
- **The instruments pattern**: policy levers toward an option (certification, procurement, PES, EPR, bans, labeling, congestion pricing, urban form) are tagged medium with explicit rationale — they drive options rather than being them.
- Unmapped options (6): fugitive CH4, nuclear, F-gas, transport biofuels, shipping/aviation, fuel-efficient vehicles (list jumps straight to zero-emission fleets).
- Unreached actions (13) are coherent: grid/T&D (no SPM.7 transmission option), pure governance instruments, cooking-fuel phase-out. Unreached = no potential evidence from this source, not unimportant.
